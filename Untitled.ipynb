```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter 
from scipy.signal import find_peaks
```

```python
dir = 'p10 7000V Cs'
with open(f"{dir}/filename.ini", "r") as f:
    file_list = list(filter(bool, f.read().split("\n")))
file_list[:5]
```

```python
arr = []
for file in file_list[:1]:
    with open(f"{dir}/{file}", "rb") as data_file:
            data = data_file.read()
            split_data = data.split(0xf0050000.to_bytes(4,"big"))
    
            byte_data = np.frombuffer(split_data[-1], dtype=np.int16)
    
            for num in byte_data:
                num = round(num/400.0 + 0, 3)
                arr.append(num)

readData = pd.DataFrame(arr)
```

```python
readData
```

```python
smootedData = readData[[0]].apply(savgol_filter,  window_length=5, polyorder=3)
smootedData2 = smootedData[[0]].apply(savgol_filter,  window_length=5, polyorder=3)
```

```python
readData.hist(bins=50)
```

```python
np.histogram(readData[[0]],bins = 50)
```

```python
sorted = readData.sort_values(0)
filtered = sorted[int(len(sorted)*0.2):int(len(sorted)*0.80)]
filtered
```

```python
std = np.std(filtered[0])
std
```

```python
avg = np.average(filtered[0])
avg
```

```python
trshld = avg + std*(2+2**0.5)
peaks, _ = find_peaks(smootedData[0],threshold = 0.020)
filtredPeaks = smootedData[0][peaks][smootedData[0]>trshld]
filtredPeaks = filtredPeaks.keys()
```

```python

plt.figure(figsize=(32, 16))
plt.plot(range(0,len(readData)),readData, color="orange", linewidth=3)

plt.axhline(avg, color = 'r', linestyle = '-')
plt.axhline(trshld, color = 'blue', linestyle = '-')

plt.plot(range(0,len(smootedData)),smootedData, color="blue", linewidth=1)
##plt.plot(range(0,len(smootedData2)),smootedData2, color="purple", linewidth=1)
plt.scatter(filtredPeaks,smootedData[0][filtredPeaks], color="red", linewidth=3)
plt.show()
```
