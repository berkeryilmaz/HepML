```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Örnek sinyal verisi oluşturma
fs = 1000  # Örnekleme frekansı (örneğin, 1000 Hz)
t = np.linspace(0, 1, fs, endpoint=False)  # 1 saniyelik bir süre içinde örnekler oluşturun
frequency = 5  # Sinyal frekansı (örneğin, 5 Hz)
signal = np.sin(2 * np.pi * frequency * t)  # Sinüsoidal sinyal oluşturun
```

```python
# FFT hesaplaması
fft_result = np.fft.fft(signal)
```

```python
# FFT hesaplaması
fft_result = np.fft.fft(signal)
```

```python
# Güç spektrumu hesaplaması
power_spectrum = np.abs(fft_result)
```

```python
# Kesme frekansını belirleme (örneğin, 50 Hz altı frekans bileşenlerini koruyalım)
cutoff_frequency = 50
power_spectrum[:cutoff_frequency] = 0
```

```python
# Ters Fourier dönüşümü ile temizlenmiş sinyal elde etme
cleaned_signal = np.fft.ifft(fft_result)
```

```python
# Sinyal ve temizlenmiş sinyal görselleştirme
plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.title('Orijinal Sinyal')
plt.subplot(2, 1, 2)
plt.plot(t, cleaned_signal)
plt.title('Temizlenmiş Sinyal')
plt.tight_layout()
plt.show()
```
