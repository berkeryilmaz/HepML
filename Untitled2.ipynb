```python
data=osc.channel[1].data

SAMPLE_RATE = 5000  # Hertz
DURATION = len(data) / SAMPLE_RATE  # Seconds
DURATION
```

```python
normalized_data = np.int16((data / max(data)) * 32767)

plt.plot(normalized_data)
plt.show()
```

```python
from scipy.fft import rfft, rfftfreq

N = len(data)

yf = rfft(normalized_data)
xf = rfftfreq(N, 1 / SAMPLE_RATE)

plt.plot(xf, np.abs(yf))
plt.show()
```

```python
yf[25:] = 0
plt.plot(xf, np.abs(yf))
plt.show()
```

```python
from scipy.fft import irfft

new_sig = irfft(yf)

plt.plot(new_sig)
plt.show()
```
